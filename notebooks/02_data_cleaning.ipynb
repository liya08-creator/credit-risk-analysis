```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("UCI_Credit_Card.csv")
df_clean = df.copy()

print(df_clean.shape)
df_clean.head()
```

```python
#  missing values
print("Missing values per column:")
print(df_clean.isnull().sum())
print("\nTotal missing values:", df_clean.isnull().sum().sum())

#  duplicates
print("\nIs ID unique?", df_clean["ID"].is_unique)
print("Fully duplicated rows (including ID):", df_clean.duplicated().sum())
print("Duplicated rows when ID is ignored:", df_clean.drop(columns="ID").duplicated().sum())
```

```python
# Rows that match another row on every column except ID
cols_no_id = df_clean.columns.drop("ID")
dups = df_clean[df_clean.duplicated(subset=cols_no_id, keep=False)]

print("Rows involved (every member of each matching group):", len(dups))
print("Number of distinct matching groups:", dups.groupby(list(cols_no_id)).ngroups)

# Sort so that matching rows sit next to each other
dups.sort_values(by=list(cols_no_id)).head(20)
```

## Cleaning Log

**1. Missing values**
Checked all 25 columns. Total missing values: 0. No action needed.

**2. Duplicates**
- ID is unique (no repeated customer IDs).
- No rows are fully identical (ID included).
- 35 pairs (70 rows) are identical on every column except ID.
- Investigation: most are inactive accounts (all-zero bills and payments, PAY status -2) with matching demographics, which is plausible by coincidence.
- Decision: KEEP. They have different IDs, cannot be proven to be the same customer, and are only about 0.1% of the data.

```python
#  undocumented category codes

valid_education = [1, 2, 3, 4]
valid_marriage = [1, 2, 3]

edu_invalid = df_clean[~df_clean["EDUCATION"].isin(valid_education)]
mar_invalid = df_clean[~df_clean["MARRIAGE"].isin(valid_marriage)]

print("EDUCATION codes outside 1-4:")
print(edu_invalid["EDUCATION"].value_counts().sort_index())
print("Total rows:", len(edu_invalid))

print("\nMARRIAGE codes outside 1-3:")
print(mar_invalid["MARRIAGE"].value_counts().sort_index())
print("Total rows:", len(mar_invalid))


target = "default.payment.next.month"
print("\nOverall default rate:", df_clean[target].mean().round(3))
print("Default rate, invalid EDUCATION rows:", edu_invalid[target].mean().round(3))
print("Default rate, invalid MARRIAGE rows:", mar_invalid[target].mean().round(3))
```

```python
# Default rate and size of every code, valid and invalid
print("EDUCATION")
print(df_clean.groupby("EDUCATION")[target].agg(["count", "mean"]).round(3))

print("\nMARRIAGE")
print(df_clean.groupby("MARRIAGE")[target].agg(["count", "mean"]).round(3))
```

```python
#  merge undocumented EDUCATION codes (0, 5, 6) into 4 ("others")
df_clean["EDUCATION"] = df_clean["EDUCATION"].replace({0: 4, 5: 4, 6: 4})

# Verify the result
print(df_clean["EDUCATION"].value_counts().sort_index())
print()
print(df_clean.groupby("EDUCATION")[target].agg(["count", "mean"]).round(3))
print()
print("MARRIAGE (left unchanged):")
print(df_clean["MARRIAGE"].value_counts().sort_index())
```

**3. Invalid / undocumented category codes**

EDUCATION
- Documented codes are 1-4. Found undocumented codes: 0 (14 rows), 5 (280 rows), 6 (51 rows). Total 345 rows (1.15%).
- Their default rates (0%, 6.4%, 15.7%) were similar to or lower than the documented "others" group (code 4: 5.7%, 123 rows).
- Decision: MERGE codes 0, 5 and 6 into 4 ("others"). Code 4 now has 468 rows.

MARRIAGE
- Documented codes are 1-3. Found undocumented code 0 (54 rows, 0.18%).
- Its default rate (9.3%) differs strongly from the documented "others" group (code 3: 26.0%), so merging would distort that group.
- Decision: KEEP code 0 as its own category ("unknown"). Rows are not removed because they cannot be proven wrong.

```python
#  repayment status columns
pay_cols = ["PAY_0", "PAY_2", "PAY_3", "PAY_4", "PAY_5", "PAY_6"]

# 1. How often does each status value appear in each column?
print(df_clean[pay_cols].apply(pd.Series.value_counts).fillna(0).astype(int))

# 2. Default rate for each status value of the most recent month (PAY_0)
print()
print(df_clean.groupby("PAY_0")[target].agg(["count", "mean"]).round(3))
```

**4. Repayment status columns (PAY_0 to PAY_6)**

- Documented values: -1 (paid on time) and 1-9 (months of delay). Found additional undocumented values: -2 and 0, which are very common (0 is the most common value in every month; -2 appears for 2,700-4,900 customers per month).
- Observed range is -2 to 8; no impossible values.
- Default rates for PAY_0: -2: 13.2%, -1: 16.8%, 0: 12.8% (low and similar), 1: 33.9%, 2: 69.1%, 3: 75.8%. The codes -2, -1 and 0 therefore behave like one group of customers paying without problems.
- -2 and 0 are commonly interpreted in the dataset's community as "no consumption" and "revolving credit / minimum payment made". This is an interpretation, not part of the official documentation.
- Notes: PAY_0 values 5-8 have very few customers (9-26) so their rates are unreliable. The value 1 is frequent in PAY_0 (3,688) but almost absent in PAY_2 to PAY_6 (0-28), a recording inconsistency that cannot be corrected from the data.
- Decision: KEEP all values unchanged. They are valid categories, and removing or merging them would lose information.

```python
#  sanity checks on numeric columns
print("AGE range:", df_clean["AGE"].min(), "to", df_clean["AGE"].max())
print("LIMIT_BAL range:", df_clean["LIMIT_BAL"].min(), "to", df_clean["LIMIT_BAL"].max())

bill_cols = [f"BILL_AMT{i}" for i in range(1, 7)]
payamt_cols = [f"PAY_AMT{i}" for i in range(1, 7)]

print("\nNegative values in BILL_AMT columns (rows):")
print((df_clean[bill_cols] < 0).sum())

print("\nNegative values in PAY_AMT columns (rows):")
print((df_clean[payamt_cols] < 0).sum())

print("\nRows with at least one negative bill:", (df_clean[bill_cols] < 0).any(axis=1).sum())
```

```python
#  how big are the negative bills, and do these customers behave differently?

neg_mask = (df_clean[bill_cols] < 0).any(axis=1)

# 1. Size of the negative values
all_neg_values = df_clean[bill_cols].stack()
all_neg_values = all_neg_values[all_neg_values < 0]
print("Negative bill values: summary")
print(all_neg_values.describe().round(0))

# 2. Default rate of customers with vs without negative bills
print("\nDefault rate, customers WITH a negative bill:", df_clean.loc[neg_mask, target].mean().round(3))
print("Default rate, customers WITHOUT:", df_clean.loc[~neg_mask, target].mean().round(3))

# 3. Do negative bills tend to be small compared with the credit limit?
worst = df_clean[bill_cols].min(axis=1)
ratio = (worst[neg_mask] / df_clean.loc[neg_mask, "LIMIT_BAL"])
print("\nMost negative bill as a share of credit limit (customers with negatives):")
print(ratio.describe().round(3))
```

**5. Numeric columns: sanity checks and negative bills**

- AGE ranges from 21 to 79 and LIMIT_BAL from 10,000 to 1,000,000. All values are plausible; no action needed.
- PAY_AMT1-6 contain no negative values.
- BILL_AMT1-6 contain negative values: 3,932 values across 1,930 customers (6.4%).
- Most negative bills are small (median -158; 75% are between -18 and 0). Negative bills are plausible: they represent credit balances from overpayment or refunds.
- A few are extreme (minimum -339,603; one customer's credit balance is 1.5x their credit limit). These cannot be proven wrong and will be reviewed with the other outliers in Activity 4.
- Customers with at least one negative bill have a lower default rate (16.5%) than the rest (22.5%), so these rows carry useful information.
- Decision: KEEP all negative bill values unchanged.

```python
#  count outliers per numeric column using the IQR rule
num_cols = ["LIMIT_BAL", "AGE"] + bill_cols + payamt_cols

rows = []
for col in num_cols:
    q1 = df_clean[col].quantile(0.25)
    q3 = df_clean[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    n_out = ((df_clean[col] < lower) | (df_clean[col] > upper)).sum()
    rows.append([col, n_out, round(100 * n_out / len(df_clean), 1), df_clean[col].min(), df_clean[col].max()])

outlier_table = pd.DataFrame(rows, columns=["column", "n_outliers", "percent", "min", "max"])
outlier_table
```

```python
#  investigate the most extreme values

# 1. Bills above the credit limit
over_limit = df_clean[bill_cols].gt(df_clean["LIMIT_BAL"], axis=0)
print("Bill values above credit limit:", over_limit.sum().sum())
print("Customers with at least one bill above limit:", over_limit.any(axis=1).sum())
print("Default rate of those customers:", df_clean.loc[over_limit.any(axis=1), target].mean().round(3))

# 2. The customer with the largest BILL_AMT3
idx_bill = df_clean["BILL_AMT3"].idxmax()
print("\nLargest BILL_AMT3 customer:")
print(df_clean.loc[[idx_bill], ["ID", "LIMIT_BAL", "BILL_AMT2", "BILL_AMT3", "BILL_AMT4", "PAY_AMT2", "PAY_AMT3", target]])

# 3. The customer with the largest PAY_AMT2
idx_pay = df_clean["PAY_AMT2"].idxmax()
print("\nLargest PAY_AMT2 customer:")
print(df_clean.loc[[idx_pay], ["ID", "LIMIT_BAL", "BILL_AMT2", "BILL_AMT3", "BILL_AMT4", "PAY_AMT2", "PAY_AMT3", target]])
```

**6. Outliers**

- IQR rule flagged: LIMIT_BAL 0.6%, AGE 0.9%, BILL_AMT1-6 about 8-9% each, PAY_AMT1-6 about 9-10% each. The high percentages in the money columns reflect right-skewed data (most customers have small amounts, a minority have very large ones), not errors.
- 8,274 bill values exceed the customer's credit limit, across 3,931 customers (13.1%). These customers have a higher default rate (30.1% vs 22.1% overall). This is plausible (interest, fees, over-limit spending) and informative.
- The largest bill (BILL_AMT3 = 1,664,089) and the largest payment (PAY_AMT2 = 1,684,259) belong to the same customer (ID 5297, limit 500,000). Their payments match the previous month's bills closely (including 121,757 billed and 121,831 paid the following month), so the extreme values are consistent, not errors.
- Decision: KEEP all outliers unchanged. They are real and informative. Skewness will be handled at the feature engineering / modelling stage (scaling, possible log transforms) instead of by deleting customers.

```python
#  check data types and whether float columns hold only whole numbers
print(df_clean.dtypes)

float_cols = df_clean.select_dtypes(include="float").columns
print("\nFloat columns:", list(float_cols))

print("\nFloat columns with non-whole values:")
for col in float_cols:
    not_whole = (df_clean[col] % 1 != 0).sum()
    print(col, ":", not_whole)
```

```python
#  convert whole-number float columns to integers
df_clean[float_cols] = df_clean[float_cols].astype("int64")

# Verify
print(df_clean.dtypes)
print()
df_clean[["LIMIT_BAL", "BILL_AMT1", "PAY_AMT1"]].head()
```

**7. Data types**

- 13 columns (LIMIT_BAL, BILL_AMT1-6, PAY_AMT1-6) were loaded as float64. Checked each: none contain fractional values (0 non-whole values in every column).
- Decision: CONVERT all 13 to int64. Amounts are whole currency units, so no information is lost, and integer types match the planned PostgreSQL schema.
- SEX, EDUCATION, MARRIAGE and PAY_0 to PAY_6 are stored as integer codes representing categories. Kept as integers for now; encoding for modelling is handled in the next step (categorical variables).

```python
#  rename columns for consistency and SQL safety
df_clean = df_clean.rename(columns={
    "PAY_0": "PAY_1",
    "default.payment.next.month": "default_next_month"
})

# Update the variables we used in earlier cells so they still point to the right columns
target = "default_next_month"
pay_cols = ["PAY_1", "PAY_2", "PAY_3", "PAY_4", "PAY_5", "PAY_6"]

print(df_clean.columns.tolist())
print(df_clean[target].value_counts())
```

**8. Column renaming**

- PAY_0 renamed to PAY_1, so repayment status columns are numbered 1-6 like BILL_AMT1-6 and PAY_AMT1-6 (PAY_1 = September, the most recent month).
- default.payment.next.month renamed to default_next_month. The original name contains dots, which are awkward in Python and invalid in SQL column names. The bare name "default" was avoided because DEFAULT is a reserved SQL keyword.
- Data values were not changed, only the names.

```python
# add readable label columns (original numeric codes are kept)
sex_map = {1: "Male", 2: "Female"}
education_map = {1: "Graduate school", 2: "University", 3: "High school", 4: "Others"}
marriage_map = {0: "Unknown", 1: "Married", 2: "Single", 3: "Others"}

df_clean["SEX_label"] = df_clean["SEX"].map(sex_map)
df_clean["EDUCATION_label"] = df_clean["EDUCATION"].map(education_map)
df_clean["MARRIAGE_label"] = df_clean["MARRIAGE"].map(marriage_map)

# Verify: no unmapped values, and counts + default rates per label
print(df_clean[["SEX_label", "EDUCATION_label", "MARRIAGE_label"]].isnull().sum())

for col in ["SEX_label", "EDUCATION_label", "MARRIAGE_label"]:
    print()
    print(df_clean.groupby(col)[target].agg(["count", "mean"]).round(3))
```

**9. Categorical variables**

- Added readable label columns next to the original codes: SEX_label (1=Male, 2=Female), EDUCATION_label (1=Graduate school, 2=University, 3=High school, 4=Others), MARRIAGE_label (0=Unknown, 1=Married, 2=Single, 3=Others).
- Verified: no unmapped values; counts and default rates match the earlier checks (e.g. Others education = 468 customers).
- Original numeric codes are kept. Labels are for analysis, SQL and Power BI readability.
- Encoding for machine learning (one-hot encoding of SEX, EDUCATION, MARRIAGE) is postponed to the modelling phase, after the train/test split, to avoid data leakage.

```python
#  check ID is irrelevant, then build the modelling dataset

# 1. Does ID relate to the target at all?
print("Correlation between ID and default:", round(df_clean["ID"].corr(df_clean[target]), 4))

# Default rate by ID block (10 equal groups of rows): should be roughly flat
id_blocks = pd.cut(df_clean["ID"], bins=10)
print()
print(df_clean.groupby(id_blocks, observed=True)[target].mean().round(3))

# 2. Build the modelling dataset
label_cols = ["SEX_label", "EDUCATION_label", "MARRIAGE_label"]
df_model = df_clean.drop(columns=["ID"] + label_cols)

print()
print("df_clean shape:", df_clean.shape)
print("df_model shape:", df_model.shape)
print(df_model.columns.tolist())
```

**10. Removing irrelevant variables from the modelling features**

- ID is only a row identifier. Correlation with the target is -0.014, and default rates are flat across ID blocks (19.1% to 25.8%, no trend). It carries no information about credit behaviour.
- Created two datasets:
  - df_clean (30,000 x 28): full cleaned dataset, with ID and label columns. Used for EDA, PostgreSQL and Power BI.
  - df_model (30,000 x 24): ID and label columns removed. 23 features + target. Used for machine learning.
- ID is kept in df_clean because it is needed as the database primary key and to link predictions back to customers in Power BI.
